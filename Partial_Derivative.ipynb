```python
import numpy as np
from scipy.stats import norm

def calculate_delta(S, K, T, r, sigma):
    """
    S: ราคาหุ้นปัจจุบัน (Spot Price)
    K: ราคาใช้สิทธิ (Strike Price)
    T: เวลาจนถึงวันหมดอายุ (ปี)
    r: อัตราดอกเบี้ยปราศจากความเสี่ยง (Risk-free rate)
    sigma: ความผันผวน (Volatility)
    """

    # คำนวณ d1 ซึ่งเป็นส่วนประกอบหลักของสูตร Black-Scholes
    d1 = (np.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))

    # Delta ของ Call Option คือ Partial Derivative ของราคาเทียบกับ S
    # ซึ่งผลลัพธ์สุทธิคือค่า Cumulative Standard Normal Distribution ของ d1
    delta_call = norm.cdf(d1)

    return delta_call

# --- ทดลองใส่ตัวเลข ---
spot_price = 100    # หุ้นราคา 100 บาท
strike_price = 105  # อยากซื้อที่ราคา 105 บาท
time_to_expiry = 0.5 # อีก 6 เดือนหมดอายุ
risk_free = 0.03    # ดอกเบี้ย 3%
volatility = 0.25   # ความผันผวน 25%

delta = calculate_delta(spot_price, strike_price, time_to_expiry, risk_free, volatility)

print(f"Delta ของ Call Option นี้คือ: {delta:.4f}")
print(f"หมายความว่า: ถ้าหุ้นแม่ขึ้น 1 บาท, Option นี้จะมีราคาสูงขึ้นประมาณ {delta:.2f} บาท")
```

2. การใช้ Gradient เพื่อหาจุดสูงสุด (Optimization)


```python
def profit_function(x):
    return -(x - 5)**2 + 100

def get_gradient(x):
    # Partial Derivative (ดิฟ) ของ -(x-5)^2 + 100 เทียบกับ x
    # จะได้ -2 * (x - 5)
    return -2 * (x - 5)

# เริ่มต้นสุ่มค่า x (เช่น การตั้งค่าพารามิเตอร์ของ Bot)
current_x = 0
learning_rate = 0.1 # ก้าวเดินทีละนิด

print("เริ่มปรับจูนพารามิเตอร์...")
for i in range(10):
    grad = get_gradient(current_x)
    # ปรับค่า x ตามทิศทางของความชัน (Gradient Ascent เพื่อหากำไรสูงสุด)
    current_x = current_x + (learning_rate * grad)
    print(f"รอบที่ {i+1}: พารามิเตอร์ x = {current_x:.4f}, กำไรคาดการณ์ = {profit_function(current_x):.4f}")

print(f"\nจุดที่ Bot ควรตั้งค่าพารามิเตอร์เพื่อให้กำไรสูงสุดคือ x ใกล้เคียง {current_x:.2f}")
```
